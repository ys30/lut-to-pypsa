# Inspect synthetic LUT-style data
These files are synthetic and do not originate from the LUT model.

```python
from pathlib import Path
from lut_to_pypsa.pipeline import load_tables, harmonize
from lut_to_pypsa.validation import validate_inputs
data_dir = Path("../data/synthetic") if Path("../data/synthetic").exists() else Path("data/synthetic")
regions, generators, demand = load_tables(data_dir)
print(validate_inputs(regions, generators, demand))
display(regions, generators.head(), demand.head())
```

```python
_, _, converted, report = harmonize(regions, generators, demand)
print(report)
display(converted.groupby("region")["p_set_mw"].agg(["min", "max", "sum"]))
```
